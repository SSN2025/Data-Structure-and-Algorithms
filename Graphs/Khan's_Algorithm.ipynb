# Kahn's Algorithm — Topological Sort

Kahn's Algorithm is a **BFS-based algorithm** used to find a topological ordering of a **Directed Acyclic Graph (DAG)**.

A topological ordering is an ordering of vertices such that for every directed edge:

`u → v`

`u` appears before `v` in the ordering.

### Key Idea

The algorithm uses the **indegree** of each vertex.

> **Indegree:** The number of edges directed towards a vertex.

A vertex with an indegree of `0` has no prerequisites, so it can be safely placed in the topological ordering.

```python
import numpy as np
from collections import deque
```

## Algorithm

1. Calculate the indegree of every vertex.
2. Insert all vertices with indegree `0` into a queue.
3. Remove a vertex from the queue.
4. Add it to the topological ordering.
5. For each outgoing edge from that vertex, decrease the indegree of the destination vertex.
6. If a destination vertex reaches indegree `0`, insert it into the queue.
7. Continue until the queue becomes empty.

If every vertex is processed, the graph is a DAG and the resulting ordering is a valid topological ordering.

If some vertices remain unprocessed, the graph contains a cycle.

```python
def khans_topo_sort(graph):
    n = graph.shape[0]
    q = deque()
    indeg = np.zeros(n)
    for v in graph:
        for i in range(v.size):
            if(v[i]==1):
                indeg[i] += 1


    for i in range(indeg.size):
        if(indeg[i]==0):
            q.append(i)

    while(q):
        x = q.popleft()
        print(x)
        for i in range(graph[x].size):
            if(graph[x][i] == 1):
                indeg[i] -=1
                graph[x][i] = 0

                if(indeg[i]==0):q.append(i)
        

    
```

## Example

We represent the directed graph using an **adjacency matrix**.

If:

`graph[u][v] == 1`

then there is a directed edge:

`u → v`

The edges in this example are:

`0 → 1`  
`0 → 2`  
`1 → 3`  
`2 → 3`  
`2 → 4`  
`3 → 5`  
`4 → 5`

```python
graph = np.array([
    # 0  1  2  3  4  5
    [0, 1, 1, 0, 0, 0],  # 0 -> 1, 2
    [0, 0, 0, 1, 0, 0],  # 1 -> 3
    [0, 0, 0, 1, 1, 0],  # 2 -> 3, 4
    [0, 0, 0, 0, 0, 1],  # 3 -> 5
    [0, 0, 0, 0, 0, 1],  # 4 -> 5
    [0, 0, 0, 0, 0, 0]   # 5
])

khans_topo_sort(graph)
```

## Result

The algorithm produces:

`0 → 1 → 2 → 3 → 4 → 5`

This is a valid topological ordering because every vertex appears before the vertices that depend on it.

### Complexity

For an adjacency-matrix representation:

- **Time:** `O(V²)`
- **Space:** `O(V)`

where `V` is the number of vertices.
