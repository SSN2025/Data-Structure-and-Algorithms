```python
import numpy as np
from collections import deque
```

```python
def khans_topo_sort(graph):
    n = graph.shape[0]
    q = deque()
    indeg = np.zeros(n)
    for v in graph:
        for i in range(v.size):
            if(v[i]==1):
                indeg[i] += 1


    for i in range(indeg.size):
        if(indeg[i]==0):
            q.append(i)

    while(q):
        x = q.popleft()
        print(x)
        for i in range(graph[x].size):
            if(graph[x][i] == 1):
                indeg[i] -=1
                graph[x][i] = 0

                if(indeg[i]==0):q.append(i)
        

    
```

```python
graph = np.array([
    # 0  1  2  3  4  5
    [0, 1, 1, 0, 0, 0],  # 0 -> 1, 2
    [0, 0, 0, 1, 0, 0],  # 1 -> 3
    [0, 0, 0, 1, 1, 0],  # 2 -> 3, 4
    [0, 0, 0, 0, 0, 1],  # 3 -> 5
    [0, 0, 0, 0, 0, 1],  # 4 -> 5
    [0, 0, 0, 0, 0, 0]   # 5
])

khans_topo_sort(graph)
```
