```python
import numpy as np
```

```python
def spnng_trs(graph): #spanning trees
    deg = np.zeros(graph[0].size)
    for i in range(graph.shape[0]):
            for j in range(graph[i].size):
                if(graph[i][j]==1):
                    deg[i]+=1
    for i in range(graph.shape[0]):
        for j in range(graph[i].size):
            if(i==j):
                graph[i][j] = deg[i]
            elif(graph[i][j] == 1):
                graph[i][j] = -1
    minor = np.zeros((graph.shape[0]-1,graph[i].size-1))
    for i in range(1,graph.shape[0]):
            for j in range(1,graph[i].size):
                 minor[i-1][j-1] = graph[i][j]
    x = int(np.linalg.det(minor))
    print(f"Total Spanning trees: {x}")
            
            
```

```python
graph = np.array([[0,0,1,1],
                  [0,0,1,1],
                  [1,1,0,1],
                  [1,1,1,0]])
spnng_trs(graph)
```
