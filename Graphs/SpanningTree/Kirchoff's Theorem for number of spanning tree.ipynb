# Kirchhoff's Matrix-Tree Theorem

Kirchhoff's Matrix-Tree Theorem is a powerful concept in graph theory used to compute the exact **number of spanning trees** in a connected graph.

To calculate this, we rely on the graph's **Laplacian matrix** ($L$), defined as:
$$L = D - A$$
Where:
*   $A$ is the adjacency matrix.
*   $D$ is the degree matrix (a diagonal matrix where each entry represents the degree of a vertex).

First, we will import the necessary numerical libraries to handle these matrix operations efficiently.

````python
import numpy as np
````

With our mathematical environment established, we can now translate the Matrix-Tree Theorem into a programmatic algorithm.

## Algorithm Implementation

According to the Matrix-Tree Theorem, the number of spanning trees equals the determinant of any **cofactor** of the Laplacian matrix $L$. 

The following function, `spnng_trs()`, automates this process:
1. Computes the degree of each vertex directly from the adjacency matrix.
2. Constructs the Laplacian matrix in place by replacing the diagonal with vertex degrees and existing edges with $-1$.
3. Computes the minor by isolating and removing the first row and first column.
4. Calculates the determinant of this minor to output the total spanning trees.

````python
def spnng_trs(graph): #spanning trees

    deg = np.sum(graph, axis=1)


    for i in range(graph.shape[0]):
        for j in range(graph[i].size):
            if(i==j):
                graph[i][j] = deg[i]
            elif(graph[i][j] == 1):
                graph[i][j] = -1
    minor = np.zeros((graph.shape[0]-1,graph[i].size-1))
    for i in range(1,graph.shape[0]):
            for j in range(1,graph[i].size):
                 minor[i-1][j-1] = graph[i][j]
    x = int(np.linalg.det(minor))
    print(f"Total Spanning trees: {x}")
            
            
````

> **Technical Note:** The `np.linalg.det()` function computes the determinant using LU factorization. We cast the final result to an integer, as the total number of valid spanning trees must always be a discrete whole number.

## Example Execution

Let's test the algorithm using a sample undirected graph. Consider a graph represented by the following adjacency matrix $A$:

$$A = \begin{bmatrix} 0 & 0 & 1 & 1 \\ 0 & 0 & 1 & 1 \\ 1 & 1 & 0 & 1 \\ 1 & 1 & 1 & 0 \end{bmatrix}$$

We will initialize this adjacency matrix as a 2D array and pass it into our function to calculate its spanning trees.

````python
graph = np.array([[0,0,1,1],
                  [0,0,1,1],
                  [1,1,0,1],
                  [1,1,1,0]])
spnng_trs(graph)
````
