# Minimum Spanning Tree using Prim's Algorithm

This notebook implements **Prim's Algorithm** to compute the Minimum Spanning Tree (MST) of a connected, undirected, weighted graph represented as an adjacency matrix.

### Overview
Prim's algorithm is a greedy algorithm that finds a minimum spanning tree for a weighted undirected graph:
1. It maintains a growing subtree, starting from the edge with the absolute minimum cost across the entire graph.
2. An array `near` tracks which tree vertex is closest to each vertex not yet added to the tree.
3. In each step, the algorithm adds the non-tree vertex with the minimum edge weight connecting it to the current tree.
4. The tracking array `near` is updated until all $n$ vertices are connected via $n - 1$ edges.

We start by importing the numpy library

```python
import numpy as np
```

## 1. Prim's Algorithm Implementation

The function `prim_algo` operates on an $n \times n$ adjacency matrix:
- **Missing Edges & Self-Loops**: Represented by `np.inf`.
- **`near` Array**:
  - `near[i] = -1`: Vertex `i` is already included in the MST.
  - `near[i] = v`: Vertex `v` is the node currently in the tree closest to vertex `i`.
- **`t` Array**: Stores the selected edges spanning the tree, resulting in an shape $(n - 1) \times 2$.

```python
def prim_algo(graph):
        #assuming that the weight of two non-connected nodes in infinity
        n = graph.shape[0]
        t = np.zeros((n-1,2))
        mincost = np.inf
        k,l=0,0
        
        for i in range(n):
                for j in range(n):
                        
                        if(i!=j and graph[i][j] < mincost):
                                mincost = graph[i][j]
                                l = j
                                k = i
        t[0][0] = k
        t[0][1] = l

        near = np.zeros(n,dtype = int)

        for i in range(n):
            near[i] = l if graph[i][l] < graph[i][k] else k

        near[k] = near[l] = -1

        for i in range(1,n-1):
            min = np.inf
            j = 0
            for l  in range(n):
                 if(near[l]!=-1 and graph[l][near[l]] < min):
                        min = graph[l][near[l]]
                        j = l
                
            t[i][0] = near[j]
            t[i][1] = j
            mincost += graph[j][near[j]]
            near[j] = -1
            for l in range(n):
                   if(near[l]!=-1 and graph[l][near[l]] > graph[l][j]):
                          near[l] = j
        print(f"Minimum cost = {mincost}")
        return t


```

## 3. Running the Algorithm

Execute the algorithm on the defined graph, displaying the total cost of the Minimum Spanning Tree and the selected edges.

```python
x = prim_algo(graph)

print(x)
```
