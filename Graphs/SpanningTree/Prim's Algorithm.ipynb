```python
def prim_algo(graph):
        #assuming that the weight of two non-connected nodes in infinity
        n = graph.shape[0]
        t = np.zeros((n-1,2))
        mincost = np.inf
        k,l=0,0
        
        for i in range(n):
                for j in range(n):
                        
                        if(i!=j and graph[i][j] < mincost):
                                mincost = graph[i][j]
                                l = j
                                k = i
        t[0][0] = l
        t[0][1] = k

        near = np.zeros(n,dtype = int)

        for i in range(n):
            near[i] = l if graph[i][l] < graph[i][k] else k

        near[k] = near[l] = -1

        for i in range(1,n-1):
            min = np.inf
            j = 0
            for l  in range(n):
                 if(near[l]!=-1 and graph[l][near[l]] < min):
                        min = graph[l][near[l]]
                        j = l
                
            t[i][0] = j
            t[i][1] = near[j]
            mincost += graph[j][near[j]]
            near[j] = -1
            for l in range(n):
                   if(near[l]!=-1 and graph[l][near[l]] > graph[l][j]):
                          near[l] = j
        print(f"Minimum cost = {mincost}")
        return t


```
