# Kruskal's Algorithm for Minimum Spanning Tree (MST)
This notebook implements Kruskal's algorithm to find the Minimum Spanning Tree of a weighted undirected graph.

```python
import numpy as np
import heapq as hq
```

## Disjoint Set Union (DSU) & Core Algorithm
This section defines the main algorithm. It utilizes a Disjoint Set Union (DSU) data structure, featuring path compression in the `find(x)` helper and union by rank in the `union(x,y)` helper to efficiently detect cycles while iterating through sorted edges.

```python
def Krushkal(n,edges):
    prnt = np.arange(n) #parent
    rnk =   np.zeros(n)  # rank

    def find(x):
        if prnt[x]!=x:
            prnt[x] = find(prnt[x])
        return prnt[x]
    def union(x,y):
        px = find(x) #parent of x
        py = find(y) #parent of y

        if rnk[px]<rnk[py]:
            prnt[px] = py
        else :
            prnt[py] = px
        if(rnk[px]==rnk[py]):rnk[px]+=1    


    edges.sort()

    t = np.zeros((n-1,2))

    mincost = 0.0
    i=-1
    for w,u,v in edges:
        j = find(u)
        k = find(v)
        if(j!=k):
            i+=1
            t[i][0] = u
            t[i][1] = v
            mincost += w
            union(j,k)
    if i < n-2:
        print("MST dosen't exist")
        return
    print(f"Minimum cost: {mincost}")
    return t 
```

## Graph Conversion Helper
The `init` function parses an adjacency matrix representation of the graph to generate a list of edges formatted as `(weight, u, v)`. It efficiently filters out missing connections, which are represented by `np.inf`.

```python
def init(graph):
    edges = []
    n = len(graph)
    for i in range(n):
        for j in range(i + 1, n):
            if graph[i][j] != np.inf:
                edges.append((graph[i][j],i,j))
    return n,edges
```
