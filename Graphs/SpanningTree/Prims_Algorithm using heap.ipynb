```python
import numpy as np
import heapq as hq
```

```python
def Prims(graph):
    n = graph.shape[0]
    edges = []
    for i in range(1,n):
        if(graph[0][i]!=np.inf):
            edges.append((graph[0][i],0,i))
    hq.heapify(edges)
    visited= set()
    visited.add(0)
    mincost = 0

    t = np.zeros((n-1,2))
    i=0
    while edges:
        w,st,end = hq.heappop(edges)
        if(end in visited):continue
        t[i][0] = st
        t[i][1] = end
        mincost += w
        visited.add(end)
        i+=1
        if(i==n-1):return t,mincost
        for j in range(n):
            if graph[end][j] != np.inf and j not in visited and j!=end:
                hq.heappush(edges,(graph[end][j], end , j))
    return t,mincost

    
```
