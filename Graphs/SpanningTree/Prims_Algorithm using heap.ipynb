# Prim's Algorithm Using a Min-Heap

## Introduction

Prim's Algorithm is a greedy algorithm used to find the **Minimum Spanning Tree (MST)** of a connected, weighted, undirected graph.

A Minimum Spanning Tree connects all vertices of the graph using exactly `V - 1` edges while minimizing the total edge weight.

In this implementation, a **min-heap** is used to efficiently select the minimum-weight edge available at each step.

The graph is represented using an **adjacency matrix**, where `np.inf` represents the absence of an edge.

### Approach

1. Start from vertex `0`.
2. Insert all edges connected to the starting vertex into a min-heap.
3. Remove the minimum-weight edge from the heap.
4. If the destination vertex has not been visited, add the edge to the MST.
5. Add the newly reached vertex's unvisited edges to the heap.
6. Continue until `V - 1` edges have been selected.

The algorithm follows the **greedy-choice property**: at every step, it chooses the minimum-weight edge that expands the current tree without forming a cycle.

````python
import numpy as np
import heapq as hq
````

## Implementation

The `Prims()` function maintains a min-heap containing candidate edges.

The heap stores edges in the form:

`(weight, start_vertex, end_vertex)`

A `visited` set prevents vertices from being added to the spanning tree more than once.

The variable `mincost` stores the total weight of the edges selected for the MST, while `t` stores the resulting MST edges.

````python
def Prims(graph):
    n = graph.shape[0]
    edges = []
    for i in range(1,n):
        if(graph[0][i]!=np.inf):
            edges.append((graph[0][i],0,i))
    hq.heapify(edges)
    visited= set()
    visited.add(0)
    mincost = 0

    t = np.zeros((n-1,2))
    i=0
    while edges:
        w,st,end = hq.heappop(edges)
        if(end in visited):continue
        t[i][0] = st
        t[i][1] = end
        mincost += w
        visited.add(end)
        i+=1
        if(i==n-1):return t,mincost
        for j in range(n):
            if graph[end][j] != np.inf and j not in visited and j!=end:
                hq.heappush(edges,(graph[end][j], end , j))
    return t,mincost

    
````

## Time and Space Complexity

Let `V` be the number of vertices and `E` be the number of edges.

### Time Complexity

The algorithm uses an adjacency matrix, so whenever a vertex is added to the MST, its entire row is scanned. Since up to `V` vertices are processed, these scans take:

`O(V²)`

Each edge can be inserted into the min-heap, and heap operations take `O(log E)` time. Therefore, heap operations contribute approximately:

`O(E log E)`

Hence, the overall time complexity of this implementation is:

**O(V² + E log E)**

For a dense graph represented by an adjacency matrix, where `E = O(V²)`, this can be expressed as:

**O(V² log V)**

### Space Complexity

The adjacency matrix itself requires:

`O(V²)`

The heap can contain up to `O(E)` candidate edges, while the `visited` set and MST require `O(V)` space.

Therefore, including the input graph, the total space complexity is:

**O(V² + E)**

If the input graph is excluded, the auxiliary space complexity is:

**O(E + V)**
