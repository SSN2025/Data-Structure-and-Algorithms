```python
# Merge Sort with NumPy

This notebook implements the **merge sort** algorithm using NumPy arrays.

## Overview

Merge sort is a recursive, divide-and-conquer sorting algorithm:

1. Divide the array into two halves.
2. Recursively sort each half.
3. Merge the sorted halves.
4. Return the fully sorted array.

The algorithm has a time complexity of **O(n log n)**.

## Implementation

- `merge(A, B)` combines two sorted arrays.
- `MergeSort(A)` recursively sorts an array.
- The final cell demonstrates the algorithm on an example array.
```

```python
def merge(A,B):
    i=0
    j=0
    n = len(A)
    m = len(B)
    C = np.zeros(m+n)
    k=0
    while(i<n and j<m):
        if(A[i] < B[j]):
            C[k] = A[i]
            i += 1
        else:
            C[k] = B[j]
            j += 1
        k += 1
    while(i<n):
        C[k] = A[i]
        k+=1
        i+=1
    while(j<m):
        C[k] = B[j]
        k+=1
        j+=1
    return C
```

```python
def MergeSort(A,start=0,end = None):
        if end is None:
              end = len(A)-1
        if(start == end):
            return np.array([A[start]])
        mid = (start+end)//2
        return merge(MergeSort(A,start,mid),MergeSort(A,mid+1,end))
```

```python
a = np.array([1,23,52,-44,2,-3,411,32,412,-4,321,23])
a = MergeSort(a)

print(a)
```
