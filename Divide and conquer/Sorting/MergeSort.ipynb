# Merge Sort Using NumPy

## Introduction

Merge Sort is a **divide-and-conquer sorting algorithm** used to arrange elements in ascending or descending order.

The algorithm works by repeatedly dividing an array into two smaller halves until each subarray contains a single element. These smaller sorted arrays are then merged to produce the final sorted array.

### Working Principle

The algorithm follows three main steps:

1. **Divide:** Split the array into two halves.
2. **Conquer:** Recursively sort both halves.
3. **Combine:** Merge the two sorted halves into one sorted array.

For an array of `n` elements, Merge Sort has a time complexity of **O(n log n)**.

This implementation uses NumPy arrays to store the input and intermediate results.

````python
import numpy as np
````

## Merge Function

The `merge(A, B)` function combines two already sorted arrays into a single sorted array.

Two pointers, `i` and `j`, are used to keep track of the current elements of `A` and `B`.

At each step, the smaller element is copied into the output array `C`.

Once all elements from one array have been processed, the remaining elements of the other array are copied directly into `C`.

Therefore, merging two arrays containing a total of `n` elements takes:

**Time Complexity: O(n)**

The temporary array `C` requires:

**Space Complexity: O(n)**

````python
def merge(A,B):
    i=0
    j=0
    n = len(A)
    m = len(B)
    C = np.zeros(m+n)
    k=0
    while(i<n and j<m):
        if(A[i] < B[j]):
            C[k] = A[i]
            i += 1
        else:
            C[k] = B[j]
            j += 1
        k += 1
    while(i<n):
        C[k] = A[i]
        k+=1
        i+=1
    while(j<m):
        C[k] = B[j]
        k+=1
        j+=1
    return C
````

## Merge Sort Function

The `MergeSort(A, start, end)` function recursively divides the input array into two halves.

The recursion continues until the subarray contains only one element. Single-element arrays are already sorted.

The two sorted halves are then passed to the `merge()` function to produce a larger sorted array.

The recurrence relation is:

`T(n) = 2T(n/2) + O(n)`

Therefore, by the Master Theorem:

**T(n) = O(n log n)**

````python
def MergeSort(A,start=0,end = None):
        if end is None:
              end = len(A)-1
        if(start == end):
            return np.array([A[start]])
        mid = (start+end)//2
        return merge(MergeSort(A,start,mid),MergeSort(A,mid+1,end))
````

## Time and Space Complexity

Let `n` be the number of elements in the input array.

### Time Complexity

At each level of recursion, all `n` elements are processed during merging.

There are `log₂(n)` levels of recursion.

Therefore:

**Time Complexity = O(n log n)**

This complexity remains the same for the best, average, and worst cases.

### Space Complexity

The `merge()` function creates a new NumPy array containing the elements of the two input arrays.

Across the recursive execution, additional arrays are created during merging. The overall auxiliary space requirement is:

**O(n)**

The recursion stack requires:

**O(log n)**

Thus, the dominant auxiliary space complexity is:

**O(n)**

````python
a = np.array([1,23,52,-44,2,-3,411,32,412,-4,321,23])
a = MergeSort(a)

print(a)
````
